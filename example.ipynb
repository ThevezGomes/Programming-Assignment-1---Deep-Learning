```python
# ==== CARREGANDO O DATASET ====

dataset = utils.Dataset("data/stage1_train")
print("Carregando dados...")
dataset.get_data()
dataset.split_data()
print("Dados carregados e divididos em treino, validação e teste.")
```

```python
# ==== TREINANDO E AVALIANDO O MODELO ====

model = utils.Model()
print("Treinando o modelo...")
model.train(dataset)
model.evaluate(dataset)
print("Treinamento e avaliação concluídos.")
```
