```python
# 1. Setup e Carregamento do Modelo via Checkpoint (Sem Retreinar)
import os
import sys

for path in ['src', '../src', '.']:
    if os.path.exists(os.path.join(path, 'utils.py')):
        abs_p = os.path.abspath(path)
        if abs_p not in sys.path:
            sys.path.append(abs_p)
        break

import src.utils as utils

device = utils.get_device()
caminho_checkpoint = 'checkpoints/checkpoint.pt' if os.path.exists('checkpoints/checkpoint.pt') else '../checkpoints/checkpoint.pt'

model = utils.carregar_checkpoint(caminho_checkpoint, device=device)
```

```python
# 2. Execução da Inferência Modularizada
# Insira o caminho de qualquer imagem abaixo:
caminho_imagem = "data/stage1_train/00071198d059ba7f5914a526d124d28e6d010c92466da21d4a04cd5413362552/images/00071198d059ba7f5914a526d124d28e6d010c92466da21d4a04cd5413362552.png"

# Recebe o caminho de uma imagem qualquer, devolve a máscara de instâncias colorida e a contagem
mascara_colorida, contagem = utils.inferir(caminho_imagem, model=model, device=device)

print(f"Contagem de instâncias detectadas: {contagem}")
```
